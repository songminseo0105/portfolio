area_code별 churn율 차이가 크지 않고, state 변수가 더 세분화된 지역 정보를 포함하므로 area_code는 모델 입력 변수에서 제외

```python
area_code_summary = (
    df_check
    .groupby("area_code")
    .agg(
        count=("churn_binary", "count"),
        churn_count=("churn_binary", "sum"),
        churn_rate=("churn_binary", "mean")
    )
    .reset_index()
)

area_code_summary["churn_rate(%)"] = area_code_summary["churn_rate"] * 100
area_code_summary = area_code_summary.sort_values("churn_rate(%)", ascending=False)

display(area_code_summary[["area_code", "count", "churn_count", "churn_rate(%)"]])

print("area_code별 churn rate 최대값:", round(area_code_summary["churn_rate(%)"].max(), 2), "%")
print("area_code별 churn rate 최소값:", round(area_code_summary["churn_rate(%)"].min(), 2), "%")
print("area_code별 churn rate 차이:", round(area_code_summary["churn_rate(%)"].max() - area_code_summary["churn_rate(%)"].min(), 2), "%p")
```

total_day_charge, total_eve_charge, total_night_charge, total_intl_charge는 각각 total_day_minutes, total_eve_minutes, total_night_minutes, total_intl_minutes와 상관계수가 거의 1에 가깝게 나타났다. 이는 charge 변수가 통화시간에 단가를 곱해 계산된 파생 변수임을 의미

```python
minute_charge_pairs = [
    ("total_day_minutes", "total_day_charge"),
    ("total_eve_minutes", "total_eve_charge"),
    ("total_night_minutes", "total_night_charge"),
    ("total_intl_minutes", "total_intl_charge")
]

corr_results = []

for minute_col, charge_col in minute_charge_pairs:
    corr = df_check[[minute_col, charge_col]].corr().iloc[0, 1]
    
    temp = df_check[df_check[minute_col] > 0].copy()
    temp["unit_rate"] = temp[charge_col] / temp[minute_col]
    
    corr_results.append({
        "minutes 변수": minute_col,
        "charge 변수": charge_col,
        "상관계수": corr,
        "평균 단가": temp["unit_rate"].mean(),
        "단가 표준편차": temp["unit_rate"].std(),
        "최소 단가": temp["unit_rate"].min(),
        "최대 단가": temp["unit_rate"].max()
    })

corr_results_df = pd.DataFrame(corr_results)

display(corr_results_df)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, StratifiedKFold, RandomizedSearchCV, cross_val_predict, ParameterSampler
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve,
    auc,
    confusion_matrix,
    classification_report
)
import warnings

from lightgbm import LGBMClassifier


try:
    from tqdm import tqdm
except ImportError:
    def tqdm(x, **kwargs):
        return x
```

```python
def find_best_threshold(y_true, proba):
    thresholds = np.arange(0.05, 0.95, 0.005)
    
    best_threshold = 0.5
    best_f1 = -1
    best_precision = 0
    best_recall = 0
    best_acc = 0
    
    result = []
    
    for th in thresholds:
        pred = (proba >= th).astype(int);
        acc = accuracy_score(y_true, pred);
        precision = precision_score(y_true, pred, zero_division=0)
        recall = recall_score(y_true, pred, zero_division=0)
        f1 = f1_score(y_true, pred, zero_division=0)
        
        result.append({
            "threshold": th,
            "accuracy": acc,
            "precision": precision,
            "recall": recall,
            "f1": f1
        })
        
        if f1 > best_f1:
            best_f1 = f1
            best_threshold = th
            best_precision = precision
            best_recall = recall
            best_acc = acc
    
    result_df = pd.DataFrame(result)
    
    return best_threshold, best_f1, best_precision, best_recall, best_acc, result_df
```

```python
target_col = "churn_binary"

X = df_model.drop(columns=["churn", "churn_binary"])
y = df_model[target_col].astype(int)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Churn 비율:")
display(y.value_counts(normalize=True).rename("ratio"))
```

```python
categorical_features = [
    "state",
    "international_plan",
    "voice_mail_plan"
]

numeric_features = [col for col in X.columns if col not in categorical_features]

print("Categorical features:", categorical_features)
print("Numeric features:", numeric_features)
```

```python
preprocess = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("num", "passthrough", numeric_features)
    ]
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("\nTrain churn ratio:")
display(y_train.value_counts(normalize=True))

print("\nTest churn ratio:")
display(y_test.value_counts(normalize=True))
```

```python
neg_count = (y_train == 0).sum()
pos_count = (y_train == 1).sum()
scale_pos_weight_base = neg_count / pos_count

print("Negative count:", neg_count)
print("Positive count:", pos_count)
print("scale_pos_weight base:", scale_pos_weight_base)
```

```python
neg_count = (y_train == 0).sum()
pos_count = (y_train == 1).sum()
scale_pos_weight_base = neg_count / pos_count

param_dist = {
    "n_estimators": [300, 500, 800, 1200, 1600, 2000],
    "learning_rate": [0.003, 0.005, 0.008, 0.01, 0.02, 0.03, 0.05],
    "num_leaves": [7, 15, 31, 63, 127, 255],
    "max_depth": [-1, 3, 4, 5, 6, 7, 8, 10, 12],
    "min_child_samples": [5, 10, 15, 20, 30, 50, 80, 100],
    "subsample": [0.65, 0.7, 0.75, 0.8, 0.85, 0.9, 1.0],
    "colsample_bytree": [0.65, 0.7, 0.75, 0.8, 0.85, 0.9, 1.0],
    "reg_alpha": [0, 0.001, 0.01, 0.05, 0.1, 0.5, 1.0, 3.0],
    "reg_lambda": [0, 0.001, 0.01, 0.05, 0.1, 0.5, 1.0, 3.0, 5.0, 10.0],
    "min_split_gain": [0, 0.001, 0.01, 0.05, 0.1],
    "scale_pos_weight": [
        1,
        scale_pos_weight_base * 0.5,
        scale_pos_weight_base * 0.75,
        scale_pos_weight_base,
        scale_pos_weight_base * 1.25,
        scale_pos_weight_base * 1.5,
        scale_pos_weight_base * 2.0
    ]
}
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

n_iter = 300

param_list = list(ParameterSampler(
    param_dist,
    n_iter=n_iter,
    random_state=42
))

tuning_results = []

best_f1 = -1
best_params = None
best_threshold = 0.5
best_oof_proba = None

for trial, params in enumerate(tqdm(param_list, desc="LightGBM tuning"), start=1):
    oof_proba = np.zeros(len(X_train))

    for fold, (tr_idx, val_idx) in enumerate(cv.split(X_train, y_train), start=1):
        X_tr = X_train.iloc[tr_idx]
        X_val = X_train.iloc[val_idx]
        y_tr = y_train.iloc[tr_idx]
        y_val = y_train.iloc[val_idx]

        model = Pipeline(steps=[
            ("preprocess", preprocess),
            ("model", LGBMClassifier(
                objective="binary",
                random_state=42,
                n_jobs=-1,
                verbose=-1,
                force_col_wise=True,
                **params
            ))
        ])

        # with warnings.catch_warnings():
        #     warnings.filterwarnings(
        #         "ignore",
        #         message="X does not have valid feature names.*",
        #         category=UserWarning
        #     )
        #     model.fit(X_tr, y_tr)
        #     oof_proba[val_idx] = model.predict_proba(X_val)[:, 1]

    threshold, f1, precision, recall, acc, threshold_result_df = find_best_threshold(
    y_train,
    oof_proba
)
    auc_score = roc_auc_score(y_train, oof_proba)

    tuning_results.append({
        "trial": trial,
        "f1": f1,
        "precision": precision,
        "recall": recall,
        "accuracy": acc,
        "auc": auc_score,
        "threshold": threshold,
        "params": params
    })

    if f1 > best_f1:
        best_f1 = f1
        best_params = params
        best_threshold = threshold
        best_oof_proba = oof_proba.copy()

        print("\n[New Best]")
        print("Trial      :", trial)
        print("F1         :", round(best_f1, 5))
        print("Precision  :", round(precision, 5))
        print("Recall     :", round(recall, 5))
        print("Accuracy   :", round(acc, 5))
        print("AUC        :", round(auc_score, 5))
        print("Threshold  :", round(best_threshold, 4))
        print("Params     :", best_params)

tuning_results_df = pd.DataFrame(tuning_results)

display(
    tuning_results_df
    .sort_values("f1", ascending=False)
    .head(10)
)

print("Best OOF F1:", best_f1)
print("Best Threshold:", best_threshold)
print("Best Params:")
display(best_params)
```
